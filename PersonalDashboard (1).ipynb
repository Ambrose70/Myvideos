```python
# import all the libraries.
import plotly.express as px
import requests
from bs4 import BeautifulSoup
import pandas as pd
import matplotlib.pyplot as plt
```

# **Time to scrape data from a website.**

```python
print("Lets scrape. This site is mainly for scraping and the data provided does not convey any real information")

# Base URLs
base_url = 'https://books.toscrape.com/catalogue/'
start_url = base_url + 'page-1.html'

# Container for scraped book data
books = []

# Scrape the first two pages
for page in range(1, 3):
    url = f'{base_url}page-{page}.html'
    response = requests.get(url)
    soup = BeautifulSoup(response.content, 'html.parser')
    articles = soup.find_all('article', class_='product_pod')

    for article in articles:
        title = article.h3.a['title']
        price = float(article.find('p', class_='price_color').text.strip().replace('£', ''))
        rating_text = article.p.get('class')[1]
        availability = article.find('p', class_='instock availability').text.strip()

        # Map textual ratings to numeric values
        rating = {'One': 1, 'Two': 2, 'Three': 3, 'Four': 4, 'Five': 5}.get(rating_text, 0)

        books.append({
            'Title': title,
            'Price': price,
            'Rating': rating,
            'Availability': availability
        })

```

# Creating a dataframe for our books

```python
   books.append({
            'Title': title,
            'Price': float(price),
            'Rating': rating,
            'Availability': availability
        })

book_df = pd.DataFrame(books)
book_df.head(20)
```

# **Creating visualization for our datasets**

```python
fig = px.box(book_df, x='Rating', y='Price', points='all',
             title='Price Distribution by Rating',
             labels={'Rating': 'Rating (1-5 Stars)', 'Price': 'Price (£)'})
fig.show()
```

For each rating from 1 to 5 stars, the box plot visualizes the distribution of book prices using five key summary statistics: the **median** (the line inside the box) represents the typical price for that rating; the **box edges** mark the 25th and 75th percentiles, showing where the middle 50% of prices fall (the interquartile range); the **whiskers** extend to show the range of prices within 1.5 times the IQR; and any **dots outside the whiskers** indicate outliers—books that are unusually cheap or expensive for their rating. Additionally, since `points='all'` is specified, every individual book price is plotted as a dot, providing a more detailed view of the distribution.


```python
avg_price = book_df.groupby('Rating', as_index=False)['Price'].mean()

fig = px.bar(avg_price, x='Rating', y='Price',
             title='Average Book Price by Rating',
             labels={'Rating': 'Rating (1-5)', 'Price': 'Avg. Price (£)'})
fig.show()
```

A bar graph displays data using rectangular bars to represent the values of different categories—in this case, book ratings from 1 to 5 stars. Each bar’s **height** corresponds to the **average price** of books within that rating group, allowing for a clear comparison across ratings. This visualization makes it easy to see trends, such as whether higher-rated books tend to be more expensive. Since each bar summarizes a group of books, it provides a quick overview of how price correlates with rating, helping to identify any patterns or anomalies in average pricing.


```python
fig = px.histogram(book_df, x='Price', nbins=40,
                   title='Distribution of Book Prices',
                   labels={'Price': 'Book Price (£)'},
                   color_discrete_sequence=['purple'])

fig.show()
```

The bar chart generated from this code shows the **average book price for each rating level (1 to 5 stars)**. By grouping the data by rating and calculating the mean price for each group, the plot allows you to quickly compare how pricing varies across different user ratings. Each bar’s height represents the **average price** of books with that specific rating, making it easy to see trends—such as whether higher-rated books tend to be more expensive or if lower-rated books are generally cheaper. This visualization helps summarize pricing patterns based on user perception or quality as implied by the ratings.
